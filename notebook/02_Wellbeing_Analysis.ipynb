```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the original dataset
df = pd.read_csv("../data/Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Basic dataset information

print("DATASET INFORMATION")
print("=" * 40)

print("Number of users:", len(df))
print("Number of features:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nTarget distribution:")
print(df["addicted_label"].value_counts())

print("\nTarget distribution (%):")
print(df["addicted_label"].value_counts(normalize=True) * 100)
```

```python
# Addiction distribution visualization

plt.figure(figsize=(7, 5))

df["addicted_label"].value_counts().sort_index().plot(
    kind="bar"
)

plt.title("Distribution of Digital Addiction")
plt.xlabel("Addiction Label")
plt.ylabel("Number of Users")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"],
    rotation=0
)

plt.tight_layout()
plt.show()
```

```python
# Screen time comparison by addiction status

screen_time = df.groupby("addicted_label")["daily_screen_time_hours"].mean()

print("Average Daily Screen Time:")
print(screen_time)

plt.figure(figsize=(7, 5))

screen_time.plot(kind="bar")

plt.title("Average Daily Screen Time by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Screen Time (Hours)")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"],
    rotation=0
)

plt.tight_layout()
plt.show()
```

```python
# Social media usage comparison by addiction status

social_media = df.groupby("addicted_label")["social_media_hours"].mean()

print("Average Social Media Usage:")
print(social_media)

plt.figure(figsize=(7, 5))

social_media.plot(kind="bar")

plt.title("Average Social Media Usage by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Social Media Usage (Hours)")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"],
    rotation=0
)

plt.tight_layout()
plt.show()
```

```python
# Gaming usage comparison by addiction status

gaming = df.groupby("addicted_label")["gaming_hours"].mean()

print("Average Gaming Usage:")
print(gaming)

plt.figure(figsize=(7, 5))

gaming.plot(kind="bar")

plt.title("Average Gaming Usage by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Gaming Usage (Hours)")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"],
    rotation=0
)

plt.tight_layout()
plt.show()
```

```python
# Sleep duration comparison by addiction status

sleep = df.groupby("addicted_label")["sleep_hours"].mean()

print("Average Sleep Duration:")
print(sleep)

plt.figure(figsize=(7, 5))

sleep.plot(kind="bar")

plt.title("Average Sleep Duration by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Sleep Duration (Hours)")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"],
    rotation=0
)

plt.tight_layout()
plt.show()
```

```python
# Stress level comparison by addiction status

# Convert stress levels into numerical values
stress_mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3
}

df["stress_level_numeric"] = df["stress_level"].map(stress_mapping)

# Calculate average stress level
stress = df.groupby("addicted_label")["stress_level_numeric"].mean()

print("Average Stress Level:")
print(stress)

# Plot
plt.figure(figsize=(8, 5))

plt.bar(
    [0, 1],
    stress.values
)

plt.title("Average Stress Level by Addiction Status")
plt.xlabel("Addiction Status")
plt.ylabel("Average Stress Level")

plt.xticks(
    [0, 1],
    ["Not Addicted", "Addicted"]
)

plt.tight_layout()
plt.show()
```

```python
# Addiction Level vs Addiction Status

addiction_status = pd.crosstab(
    df["addiction_level"],
    df["addicted_label"]
)

print("Addiction Level vs Addiction Status:")
print(addiction_status)

plt.figure(figsize=(8, 5))

addiction_status.plot(
    kind="bar",
    ax=plt.gca()
)

plt.title("Addiction Level by Addiction Status")
plt.xlabel("Addiction Level")
plt.ylabel("Number of Users")

plt.xticks(rotation=0)

plt.legend(
    ["Not Addicted", "Addicted"],
    title="Addiction Status"
)

plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis

# Convert Addiction Level into numeric values
df["addiction_level_numeric"] = df["addiction_level"].map({
    "Mild": 1,
    "Moderate": 2,
    "Severe": 3
})

# Select only numeric columns
numeric_columns = df.select_dtypes(include="number").columns

# Calculate correlation
correlation = df[numeric_columns].corr()

print("Correlation with Addiction Level:")
print(correlation["addiction_level_numeric"].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

correlation_with_addiction = correlation["addiction_level_numeric"].drop(
    "addiction_level_numeric"
).sort_values(ascending=False)

plt.figure(figsize=(10, 6))

plt.bar(
    correlation_with_addiction.index,
    correlation_with_addiction.values
)

plt.title("Correlation of Factors with Addiction Level")
plt.xlabel("Factors")
plt.ylabel("Correlation")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```
